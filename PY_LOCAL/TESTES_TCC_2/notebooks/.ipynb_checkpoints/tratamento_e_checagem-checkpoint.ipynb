```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
import statsmodels as sm

nomes_colunas = [
    'cod_municipio', 
    'nome_municipio',
    'perc_doc_sobrecarga',          
    'taxa_informal', 
    'taxa_climatizacao', 
    'taxa_internet',
    'taxa_lab',
    'media_alunos_por_sala',
    'renda_pc',
    'evasao',
    'tx_repet' 
]

pasta_dados = Path('C:\\Users\\patri\\PY_LOCAL\\TESTES_TCC_2\\data')
caminho_arquivo = pasta_dados / 'TESTE_ML_COM_TAXAS_INFRA.csv'
df = pd.read_csv(caminho_arquivo, sep=';', header=None, names=nomes_colunas)

display(df)
```

```python
colunas_df = ['perc_doc_sobrecarga', 'taxa_informal', 'taxa_climatizacao', 'taxa_internet', 'taxa_lab','media_alunos_por_sala',
             'renda_pc', 'tx_repet']
colunas_X = df[colunas_df]
X = colunas_X
y = df['evasao']

model = LinearRegression()
model.fit(X,y)
y_pred = model.predict(X)

residuos = y - y_pred

plt.figure(figsize=(8, 5))
sns.regplot(
    x=y_pred, 
    y=residuos, 
    lowess=True, 
    scatter_kws={'color': 'blue', 'alpha': 0.5}, 
    line_kws={'color': 'red', 'lw': 2}
)
plt.axhline(0, color='black', linestyle='--')
plt.xlabel("Valores Ajustados ($\hat{y}$)")
plt.ylabel("Resíduos ($y - \hat{y}$)")
plt.title("Análise de Resíduos Globais - Modelo Múltiplo de Evasão")
plt.tight_layout()
plt.show()
```

```python
import dcor

# 1. Selecionar colunas numéricas e remover colunas de identificação
numericas = df.select_dtypes(include='number').drop(columns=['cod_municipio'], errors='ignore')

# Tratar NaN: o dcor precisa de vetores sem valores ausentes para o cálculo
numericas_clean = numericas.dropna()

# 2. Inicializar DataFrame para a matriz de correlação de distância
cols = numericas_clean.columns
corr_dcor = pd.DataFrame(index=cols, columns=cols, dtype=float)

# 3. Preencher a matriz com o cálculo de Distance Correlation par a par
for col1 in cols:
    for col2 in cols:
        if col1 == col2:
            corr_dcor.loc[col1, col2] = 1.0
        else:
            corr_dcor.loc[col1, col2] = dcor.distance_correlation(
                numericas_clean[col1].values, 
                numericas_clean[col2].values
            )

# 4. Gerar o Heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(
    corr_dcor, 
    annot=True, 
    fmt='.2f', 
    cmap='Blues',   # Mapa de cores sequencial (melhor para a escala de 0 a 1)
    vmin=0, 
    vmax=1, 
    square=True
)
plt.title('Matriz de Correlação de Distância (dcor)')
plt.tight_layout()
plt.show()
```

```python
numericas = df.select_dtypes(include='number').drop(columns=['cod_municipio'])

corr_spearman = numericas.corr(method='spearman')

plt.figure(figsize=(9, 7))
sns.heatmap(corr_spearman, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            vmin=-1, vmax=1, square=True)
plt.title('Matriz de correlação de Spearman')
plt.tight_layout()
plt.show()
```
