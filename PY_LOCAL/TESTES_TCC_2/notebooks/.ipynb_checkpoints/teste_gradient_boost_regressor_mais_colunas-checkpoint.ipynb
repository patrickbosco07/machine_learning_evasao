```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV, KFold
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from pathlib import Path
import matplotlib.pyplot as plt
import sys

nomes_colunas = [
    'cod_municipio', 
    'nome_municipio',
    'perc_doc_sobrecarga',          
    'taxa_informal', 
    'taxa_climatizacao', 
    'taxa_internet',
    'taxa_lab',
    'media_alunos_por_sala',
    'renda_pc',
    'evasao',
    'tx_repet' 
]

pasta_dados = Path('C:\\Users\\patri\\PY_LOCAL\\TESTES_TCC_2\\data')
caminho_arquivo = pasta_dados / 'TESTE_ML_COM_TAXAS_INFRA.csv'
df = pd.read_csv(caminho_arquivo, sep=';', header=None, names=nomes_colunas)

display(df)
```

```python
colunas_ajuste = ['taxa_informal','tx_repet','evasao', 'renda_pc','taxa_internet', 'taxa_climatizacao','taxa_lab','media_alunos_por_sala']

for col in colunas_ajuste:
    df[col] = df[col].astype(str).str.replace(',', '.').astype(float)
    df[col] = df[col].round(2)

display(df)
```

```python
colunas_ml = ['perc_doc_sobrecarga','taxa_informal','tx_repet','renda_pc',
              'taxa_internet', 'taxa_climatizacao','taxa_lab','media_alunos_por_sala']

x = df[colunas_ml]
y = df['evasao']

x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.3, random_state=42)

modelo = GradientBoostingRegressor(random_state=42, loss='absolute_error')

parametrod_grid = {
    'n_estimators':     [100, 200, 300],
    'learning_rate':    [0.01, 0.05, 0.1],
    'max_depth':        [2, 3, 4],
    'min_samples_leaf': [5, 10, 20],
    'max_features':     [None, 'sqrt'],
    'subsample':        [0.7, 0.9, 1.0]
}

cv = KFold(n_splits=5, shuffle=True, random_state=42)

grid_search = GridSearchCV(estimator=modelo, param_grid=parametrod_grid,
            cv=cv, scoring='neg_mean_absolute_error', verbose=1, n_jobs=-1)

grid_search.fit(x_treino, y_treino)

print("Melhores Hiperparâmetros encontradas:")
print(grid_search.best_params_)

print(f"\nMelhor pontuação média (MAE na validação cruzada): {grid_search.best_score_:.4f}")
```

```python
melhor_modelo = grid_search.best_estimator_

melhor_modelo.fit(x_treino, y_treino)

# Faz a previsão direto no conjunto de teste isolado
previsoes = melhor_modelo.predict(x_teste)

mae = mean_absolute_error(y_teste, previsoes)
r2 = r2_score(y_teste, previsoes)

print(f"\nMAE final: {mae:.4f}")
print(f"\nR2 final: {r2:.4f}")
```

```python

importancias = melhor_modelo.feature_importances_

importancias_df = pd.Series(importancias, index=x_treino.columns).sort_values(ascending=True)
plt.figure(figsize=(10, 6))
importancias_df.plot(kind='barh', color='teal')
plt.title('Importância das Variáveis - GradientBoostingRegressor')
plt.xlabel('Importância Relativa')
plt.ylabel('Variáveis')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
from sklearn.inspection import permutation_importance

resultado = permutation_importance(
    melhor_modelo, x_teste, y_teste, n_repeats=10, random_state=42
)

importancia_perm_df = pd.Series(
    resultado.importances_mean, index=x_teste.columns
).sort_values(ascending=True)

plt.figure(figsize=(10, 6))
importancia_perm_df.plot(kind='barh', color='teal')
plt.title('Permutation Importance - GradientBoostingRegressor')
plt.xlabel('Redução Média no Desempenho')
plt.ylabel('Variáveis')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
