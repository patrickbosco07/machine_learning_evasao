```python
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from pathlib import Path
import matplotlib.pyplot as plt
from sklearn.metrics import (
    f1_score,
    accuracy_score,
    precision_score,
    recall_score,
    classification_report
)
from sklearn.inspection import permutation_importance

nomes_colunas = [
    'cod_municipio', 
    'nome_municipio',
    'perc_doc_sobrecarga',          
    'taxa_informal', 
    'taxa_climatizacao', 
    'taxa_internet',
    'taxa_lab',
    'media_alunos_por_sala',
    'renda_pc',
    'evasao',
    'tx_repet' 
]

pasta_dados = Path('C:\\Users\\patri\\PY_LOCAL\\TESTES_TCC_2\\data')
caminho_arquivo = pasta_dados / 'TESTE_ML_COM_TAXAS_INFRA.csv'
df = pd.read_csv(caminho_arquivo, sep=';', header=None, names=nomes_colunas)

colunas_ajustadas = ['taxa_informal','tx_repet','evasao', 'renda_pc','taxa_internet']

for col in colunas_ajustadas:
    df[col] = df[col].astype(str).str.replace(',', '.').astype(float)
    df[col] = df[col].round(2)

# Criar a coluna 'enquadro'
df['enquadro'] = pd.NA

# evasao entre 0 e 5 -> enquadro = 0
df.loc[df['evasao'].between(0, 5.99), 'enquadro'] = 0

# evasao maior ou igual a 6 -> enquadro = 1
df.loc[df['evasao'] >= 6, 'enquadro'] = 1

# Converter para inteiro
df['enquadro'] = df['enquadro'].astype('Int64')

display(df.head())

display(df)

display(df.isna().sum())
```

```python
colunas_ml = ['perc_doc_sobrecarga','taxa_informal','tx_repet','renda_pc',
              'taxa_internet']

x = df[colunas_ml]
y = df['enquadro']

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Monta o pipeline: SMOTE → Classificador
pipeline = ImbPipeline([
    ('smote', SMOTE(random_state=42)),
    ('clf',   GradientBoostingClassifier(random_state=42))
])

# Parâmetros com prefixo do step ('smote__' e 'clf__')
param_grid = {
    'smote__k_neighbors':    [3, 5],           # hiperparâmetro do SMOTE
    'clf__n_estimators':     [100, 200],
    'clf__learning_rate':    [0.05, 0.1],
    'clf__max_depth':        [2, 3],
    'clf__min_samples_leaf': [5, 10],
    'clf__subsample':        [0.8, 1.0],
}

# StratifiedKFold garante proporção das classes em cada fold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=cv,
    scoring='f1',       # F1 é mais adequado que accuracy para dados desbalanceados
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)

print("Melhores parâmetros:", grid.best_params_)
print(f"Melhor F1 (CV 5-fold): {grid.best_score_:.4f}")
```

```python
melhor_modelo = grid.best_estimator_

melhor_modelo.fit(X_train, y_train)

# Predições no conjunto de teste
y_pred = melhor_modelo.predict(X_test)

# Métricas finais
metricas_finais = pd.DataFrame({
    'Métrica': [
        'F1-Score',
        'Acurácia',
        'Precisão',
        'Recall'
    ],
    'Valor': [
        f1_score(y_test, y_pred, zero_division=0),
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred, zero_division=0),
        recall_score(y_test, y_pred, zero_division=0)
    ]
})

metricas_finais['Valor'] = metricas_finais['Valor'].round(4)

print('\nMétricas finais:')
display(metricas_finais)

permutation = permutation_importance(
    melhor_modelo,
    X_test,
    y_test,
    scoring='f1',
    n_repeats=30,
    random_state=42,
    n_jobs=-1
)

tabela_permutation_importance = pd.DataFrame({
    'Variável': X_test.columns,
    'Importância média': permutation.importances_mean,
    'Desvio padrão': permutation.importances_std
}).sort_values(
    by='Importância média',
    ascending=False
).reset_index(drop=True)

tabela_permutation_importance['Importância média'] = (
    tabela_permutation_importance['Importância média'].round(4)
)

tabela_permutation_importance['Desvio padrão'] = (
    tabela_permutation_importance['Desvio padrão'].round(4)
)

print('\nTabela de Permutation Importance:')
display(tabela_permutation_importance)
```
